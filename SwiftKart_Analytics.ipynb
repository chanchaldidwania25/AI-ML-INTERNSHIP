```python
df['Payment_Method']=df['Payment_Method'].str.strip().str.upper()
```

```python
df.groupby('Category')['Order_Value'].transform('median')
```

```python
df['Delivery_Delay']=df['Actual_Delivery_Days']-df['Est_Delivery_Days']
```

```python
df
```

```python
condition=[df['Delivery_Delay']<0,df['Delivery_Delay']==0,(df['Delivery_Delay']<0) & (df['Delivery_Delay']<=2), df['Delivery_Delay']>2]
Choice=['Early','On-Time','Minor Delay','Severe Delay']
df['Fulfillment_Status']=np.select(condition,Choice,default='Unknown Status')
```

```python
df
```

```python
threshold=np.percentile(df['Order_Value'],95)
df['High_Value_Order']=np.where(df['Order_Value']>threshold,'High Value Order','Normal Order')
```

```python
df
```

```python
df['Net_Revenue']=np.round(df['Order_Value']*(1-df['Discount_Percent']/100),2)
```

```python
df
```

```python
df['Category'].value_counts()
```

```python
total=df.groupby('Category')['Net_Revenue'].sum()
total
```

```python
severe_delay_pct = (
    df.groupby('City_Tier')['Fulfillment_Status']
      .apply(lambda x: (x == "Severe Delay").mean() * 100)
      .sort_values(ascending=False))
severe_delay_pct
```

```python
median=df['Order_Value'].median()
```

```python
condition=[(df['Discount_Percent']>=30) & (df['Returned']=='Yes'),(df['Order_Value']>median) & (df['Customer_Rating']>=4) & (df['Returned']=='No')]
Choice=['High Risk Transaction','Loyal & Satisfied']
df['Customer_Segmentation']=np.select(condition,Choice,default='Standard Order')
```

```python
df
```

```python
pivot=pd.pivot_table(df,values='Customer_Rating',index='Category',columns='Fulfillment_Status',aggfunc='mean')
pivot=pivot.reindex(columns=['Early', 'On-Time' , 'Minor Delay' , 'Severe Delay' ])
print(pivot)
```

```python
return_rate = df.groupby('Category')['Returned'].apply(lambda x: (x == 'Yes').mean() * 100)

print(return_rate)
```

```python
df["Category"] = df["Category"].str.strip()
total_revenue = df.groupby("Category")["Net_Revenue"].sum().sort_values(ascending=False)

plt.figure(figsize=(8,6))
total_revenue.plot(kind="bar", color=["#1f77b4","#ff7f0e","#2ca02c","#d62728","#9467bd"])
plt.title("Total Net Revenue by Category")
plt.ylabel("Revenue (INR)")
plt.xlabel("Category")
plt.grid(alpha=0.3, linestyle="--")
plt.xticks(rotation=45)
plt.gca().set_yticklabels([f"₹{int(y):,}" for y in plt.gca().get_yticks()])
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,6))

plt.hist(df["Actual_Delivery_Days"], bins=range(1,9),
         color="#1f77b4", alpha=0.7, edgecolor="black", rwidth=0.9)

plt.axvline(df["Est_Delivery_Days"].mean(), color="red", linestyle="--", linewidth=2, label="Estimated Benchmark")
plt.axvline(df["Actual_Delivery_Days"].mean(), color="green", linestyle="--", linewidth=2, label="Actual Mean")

plt.title("Distribution of Actual Delivery Days")
plt.xlabel("Delivery Days")
plt.ylabel("Frequency")
plt.xticks(range(1,8))
plt.legend()
plt.grid(alpha=0.3, linestyle="--")
plt.tight_layout()
plt.show()
```

```python
colors = df["Returned"].map({"Yes":"red", "No":"green"})

plt.figure(figsize=(8,6))
plt.scatter(df["Discount_Percent"], df["Order_Value"], c=colors, alpha=0.6, edgecolor="black")
plt.title("Discount vs Order Value (Return Status)")
plt.xlabel("Discount Percent (%)")
plt.ylabel("Order Value (INR)")
plt.grid(alpha=0.3, linestyle="--")
plt.tight_layout()
plt.show()
```

```python
rating_by_status = df.groupby("Fulfillment_Status")["Customer_Rating"].mean()

plt.figure(figsize=(8,6))
rating_by_status.plot(kind="bar", color=["#1f77b4","#ff7f0e","#2ca02c","#d62728","#9467bd"])
plt.title("Average Customer Rating by Fulfillment Status")
plt.xlabel("Fulfillment Status")
plt.ylabel("Average Rating")
plt.grid(alpha=0.3, linestyle="--")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
